```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Sample numerical dataset
data = pd.DataFrame({
    'Feature1': [2.5, 0.5, 2.2, 1.9, 3.1, 2.3, 2.0, 1.0, 1.5, 1.1],
    'Feature2': [2.4, 0.7, 2.9, 2.2, 3.0, 2.7, 1.6, 1.1, 1.6, 0.9],
    'Feature3': [1.1, 0.3, 1.5, 1.3, 1.8, 1.6, 1.0, 0.5, 1.2, 0.6]
})

print("Original Dataset:")
print(data)
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)

print("\nStandardized Data:")
print(scaled_data)
```

```python
pca_full = PCA()
pca_full.fit(scaled_data)

explained_variance = pca_full.explained_variance_ratio_

print("\nExplained Variance Ratio:")
print(explained_variance)
```

```python
plt.figure()
plt.plot(
    range(1, len(explained_variance) + 1),
    explained_variance,
    marker='o'
)
plt.xlabel('Principal Components')
plt.ylabel('Explained Variance Ratio')
plt.title('Scree Plot (Individual Variance)')
plt.grid()
plt.show()
```

```python
plt.figure()
plt.plot(
    np.cumsum(explained_variance),
    marker='o'
)
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Scree Plot (Cumulative Variance)')
plt.grid()
plt.show()
```

```python
pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)

print("\nData after PCA (2 Components):")
print(pca_data)
```

```python
plt.figure()
plt.scatter(pca_data[:, 0], pca_data[:, 1])
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('2D Visualization using PCA')
plt.grid()
plt.show()
```
